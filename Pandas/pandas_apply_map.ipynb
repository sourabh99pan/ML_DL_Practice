```python
import pandas as pd
df = pd.DataFrame({
    "Name": [" Amit ", "RAHUL", " Priya", "neha "],
    "Age": [25, 32, 28, 35],
    "Department": ["QA", "Dev", "QA", "Dev"]
})
```

```python
df["Name"] = df["Name"].str.strip()
print(df)
```

```python
df["Name"] = df["Name"].str.lower()
print(df)
```

```python
df["NameLength"] = df["Name"].str.len()
print(df)
```

```python
df["AgeGroup"] = df["Age"].apply(lambda x: "Senior" if x>=30 else "Junior")
print(df)
```

```python
df["Department"] =df["Department"].map({"QA": "Quality Assurance", "Dev": "Development"})
print(df)
```

```python
def group(age):

    if age>=30:
       return "Senior"
    else:
       return "Junior"

df["AgeGroup"] = df["Age"].apply(group)

print(df)
```
