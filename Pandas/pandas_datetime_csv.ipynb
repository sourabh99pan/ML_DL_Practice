```python
import pandas as pd
df = pd.DataFrame({
    "Name": ["Amit", "Rahul", "Priya"],
    "JoiningDate": ["2022-05-10", "2023-08-15", "2021-12-20"]
})
```

```python
df["JoiningDate"] = pd.to_datetime(df["JoiningDate"])
```

```python
df["Year"] = df["JoiningDate"].dt.year
print(df)
```

```python
df["Month"] = df["JoiningDate"].dt.month
print(df)
```

```python
df["Day"] = df["JoiningDate"].dt.day
print(df)
```

```python
df["DayOfWeek"] = df["JoiningDate"].dt.day_name()
print(df)
```

```python
df["MonthName"] = df["JoiningDate"].dt.month_name()
print(df)
```

```python
df["Quarter"] = df["JoiningDate"].dt.quarter
print(df)
```

```python
df["IsWeekend"] = df["JoiningDate"].dt.dayofweek>=5
print(df)
```
