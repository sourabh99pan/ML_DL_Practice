```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": ["Amit", "Rahul", "Priya", "Neha", "Karan", "Rahul"],
    "Age": [25, 32, 28, 35, np.nan, 32],
    "Department": ["QA", "Dev", "QA", "Dev", "QA", "Dev"],
    "Salary": [40000, 65000, 55000, 75000, np.nan, 65000],
    "City": ["Indore", "Mumbai", "Delhi", "Indore", "Pune", "Mumbai"]
})
```

```python
m = np.isnan(df["Age"]).sum()
print(m)
m = np.isnan(df["Salary"]).sum()
print(m)
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
print(df)
```

```python
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())
print(df)
```

```python
filter = df[(df["Age"]>=30) & (df["Salary"]>60000)]
print(filter["Name"])
```

```python
df.sort_values("Salary", ascending=False)
```

```python
df["AnnualSalary"] = df["Salary"]*12

print(df)
```

```python
r=df.groupby("Department")["Salary"].mean()
print(r.idxmax())
```

```python
res = df.groupby("Department").agg(Min_sal=("Salary","min"), Max_sal = ("Salary", "max"), Avg_sal=("Salary", "mean"), Num_emp=("Name", "count"))

print(res)
```

```python
df["AgeGroup"] = df["Age"].apply(lambda x: "Senior" if x>=30 else "Junior")
print(df)
```
