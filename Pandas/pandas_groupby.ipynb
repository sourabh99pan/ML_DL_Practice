```python
import pandas as pd

data = {
    "Name": ["Amit", "Rahul", "Priya", "Neha", "Karan", "Vikas"],
    "Department": ["QA", "Dev", "QA", "Dev", "QA", "Dev"],
    "Salary": [40000, 65000, 55000, 75000, 38000, 70000],
    "Experience": [2, 5, 4, 7, 1, 6]
}

df = pd.DataFrame(data)
print(df)
```

```python
df.groupby("Department")["Salary"].mean()
```

```python
df.groupby("Department")["Experience"].sum()
```

```python
df.groupby("Department")["Name"].count()
```

```python
df.groupby("Department")["Salary"].agg(
    ["min", "max", "mean"]
)
```

```python
df1=df.groupby("Department")["Salary"].mean()
print(df1.idxmax())
```

```python
data = {
    "Name": ["Amit", "Rahul", "Priya", "Neha", "Karan", "Vikas", "Riya"],
    "Department": ["QA", "Dev", "QA", "Dev", "QA", "Dev", "QA"],
    "Salary": [40000, 65000, 55000, 75000, 38000, 70000, 60000],
    "Experience": [2, 5, 4, 7, 1, 6, 5]
}

df2 = pd.DataFrame(data)
```

```python
df3=df2.groupby("Department")["Salary"].agg(
    ["min", "max", "mean"])
print(df3)
```

```python
df3.sort_values(by="mean", ascending=False)
```
