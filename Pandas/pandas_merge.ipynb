```python
import pandas as pd
employees = pd.DataFrame({
    "EmployeeID": [101, 102, 103, 104],
    "Name": ["Amit", "Rahul", "Priya", "Neha"],
    "DeptID": [1, 2, 1, 4]
})
print(employees)
departments = pd.DataFrame({
    "DepartmentID": [1, 2, 3],
    "Department": ["QA", "Development", "HR"]
})
print(departments)
```

```python
result = pd.merge(employees, departments,left_on="DeptID", right_on="DepartmentID",how="inner")
print(result)
```

```python
result = pd.merge(employees, departments,left_on="DeptID", right_on="DepartmentID",how="left")
print(result)
```

```python
result = pd.merge(employees, departments,left_on="DeptID", right_on="DepartmentID",how="outer")
print(result)
```

```python
customers = pd.DataFrame({
    "CustomerID": [101, 102, 103],
    "Name": ["Amit", "Rahul", "Priya"]
})

orders = pd.DataFrame({
    "CustomerID": [101, 101, 102, 103, 103],
    "Amount": [5000, 3000, 7000, 2000, 4000]
})
print(customers)
print(orders)
```

```python
result=pd.merge(customers, orders, on="CustomerID", how="inner")
print(result)
```

```python
total_spend = result.groupby("CustomerID")["Amount"].sum()
print(total_spend)
```

```python
avg_spend = result.groupby("CustomerID")["Amount"].mean()
print(avg_spend)
```

```python
result["total_spend"]=result.groupby("CustomerID")["Amount"].sum()
print(result)
```

```python
result["AverageOrder"]=result.groupby("CustomerID")["Amount"].mean()
print(result)
```
