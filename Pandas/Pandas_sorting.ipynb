```python
import pandas as pd
data = {
    "Name": ["Amit", "Rahul", "Priya", "Neha", "Karan"],
    "Age": [25, 32, 28, 35, 24],
    "Salary": [40000, 65000, 55000, 75000, 38000],
    "City": ["Indore", "Mumbai", "Delhi", "Indore", "Pune"]
}

df = pd.DataFrame(data)
print(df)
```

```python
df.sort_values("Salary")
```

```python
df.sort_values("Salary", ascending=False)
```

```python
df.sort_values(
    ["City", "Salary"],
    ascending=[True, False]
)
```

```python
df["AnnualSalary"] = df["Salary"] * 12
print(df)
```

```python
df["IsSenior"] = df["Age"] >= 30
print(df)
```
