```python
X_train = X_train.drop("Id", axis=1)
X_test = X_test.drop("Id", axis=1)
```

```python
X_train["HouseAge"] = X_train["YrSold"] - X_train["YearBuilt"]
X_test["HouseAge"] = X_test["YrSold"] - X_test["YearBuilt"]
```

```python
X_train["TotalBath"] = (
    X_train["FullBath"]
    + 0.5 * X_train["HalfBath"]
    + X_train["BsmtFullBath"]
    + 0.5 * X_train["BsmtHalfBath"]
)

X_test["TotalBath"] = (
    X_test["FullBath"]
    + 0.5 * X_test["HalfBath"]
    + X_test["BsmtFullBath"]
    + 0.5 * X_test["BsmtHalfBath"]
)
```

```python
X_train["TotalSF"] = (
    X_train["TotalBsmtSF"]
    + X_train["1stFlrSF"]
    + X_train["2ndFlrSF"]
)

X_test["TotalSF"] = (
    X_test["TotalBsmtSF"]
    + X_test["1stFlrSF"]
    + X_test["2ndFlrSF"]
)
```

```python
import numpy as np

X_train["GarageAreaPerCar"] = np.where(
    X_train["GarageCars"] > 0,
    X_train["GarageArea"] / X_train["GarageCars"],
    0
)

X_test["GarageAreaPerCar"] = np.where(
    X_test["GarageCars"] > 0,
    X_test["GarageArea"] / X_test["GarageCars"],
    0
)
```

```python
num_cols = X_train.select_dtypes(
    include=["int64", "float64"]
).columns

cat_cols = X_train.select_dtypes(
    include=["object"]
).columns
```

```python
from sklearn.preprocessing import OneHotEncoder

categorical_transformer = OneHotEncoder(
    drop="first",
    handle_unknown="ignore"
)
```

```python
from sklearn.compose import ColumnTransformer
preprocessor = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(
            drop="first",
            handle_unknown="ignore"
        ), cat_cols)
    ],
    remainder="passthrough"
)
```

```python
X_train_processed = preprocessor.fit_transform(X_train)
```

```python
#transform only as it protects data leakage
X_test_processed = preprocessor.transform(X_test)
```

```python
print(X_train_processed.shape)
print(X_test_processed.shape)
```

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
#don't specify max_depth This means we're allowing the tree to grow freely. 
# We actually want to see whether it overfits.
tree_model = DecisionTreeRegressor(
    random_state=42
)
```

```python
tree_model.fit(
    X_train_processed,
    y_train
)
```

```python
y_train_pred_tree = tree_model.predict(X_train_processed)

y_test_pred_tree = tree_model.predict(X_test_processed)
```

```python
from sklearn.metrics import r2_score

train_r2_tree = r2_score(
    y_train,
    y_train_pred_tree
)

test_r2_tree = r2_score(
    y_test,
    y_test_pred_tree
)

print("Decision Tree Training R²:", train_r2_tree)
print("Decision Tree Test R²:", test_r2_tree)
```

```python

for depth in [2, 3, 5, 7, 10, 15]:
    tree = DecisionTreeRegressor(
        max_depth=depth,
        random_state=42
    )

    tree.fit(X_train_processed, y_train)

    train_pred = tree.predict(X_train_processed)
    test_pred = tree.predict(X_test_processed)

    print(
        "Depth:", depth,
        "Train R²:", r2_score(y_train, train_pred),
        "Test R²:", r2_score(y_test, test_pred)
    )
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.tree import DecisionTreeRegressor

tree = DecisionTreeRegressor(random_state=42)

param_grid = {
    "max_depth": [2, 3, 5, 7, 10, 15, 20],
    "min_samples_split": [2, 5, 10],
    "min_samples_leaf": [1, 2, 5]
}

tree_grid = GridSearchCV(
    tree,
    param_grid,
    cv=5,
    scoring="r2",
    n_jobs=-1
)

tree_grid.fit(X_train_processed, y_train)
```

```python
print("Best Parameters:", tree_grid.best_params_)
print("Best CV R²:", tree_grid.best_score_)
```

```python
best_tree = tree_grid.best_estimator_

y_test_pred_tree = best_tree.predict(X_test_processed)

tree_test_r2 = r2_score(y_test, y_test_pred_tree)

print("Final Decision Tree Test R²:", tree_test_r2)
```
