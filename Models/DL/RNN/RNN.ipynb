```python
import numpy as np
import tensorflow as tf

from tensorflow.keras import Sequential
from tensorflow.keras.layers import Input, SimpleRNN, Dense
```

```python
X = []
y = []

for i in range(1000):
    X.append([[i], [i + 1], [i + 2]])
    y.append(i + 3)

X = np.array(X, dtype="float32")
y = np.array(y, dtype="float32")
print(X.shape)
print(y.shape)
```

```python
X_scaled = X / 1000.0
y_scaled = y / 1000.0
```

```python
model_rnn = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(3, 1)),
    tf.keras.layers.SimpleRNN(16, activation="tanh"),
    tf.keras.layers.Dense(1)
])

model_rnn.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="mse",
    metrics=["mae"]
)
```

```python
history_rnn = model_rnn.fit(
    X_scaled,
    y_scaled,
    epochs=50,
    batch_size=32,
    validation_split=0.2,
    verbose=1
)
```

```python
new_sequence = np.array([
    [[6], [7], [8]]
], dtype="float32")

new_sequence_scaled = new_sequence / 1000.0

prediction = model_rnn.predict(new_sequence_scaled)

print("Predicted next value:", prediction[0][0] * 1000)
```

```python
predictions = model_rnn.predict(X_scaled)

predictions_original = predictions.flatten() * 1000

for actual, predicted in zip(y[:10], predictions_original[:10]):
    print(f"Actual: {actual:.0f}, Predicted: {predicted:.2f}")
```
