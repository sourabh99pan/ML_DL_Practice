```python
import numpy as np

# Input and actual output
x = 2.0
y = 1.0

# Hidden layer
w1 = 0.5
b1 = 0.0

# Output layer
w2 = 1.0
b2 = 0.0

# Learning rate
lr = 0.1
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

```python
#Notice we're passing the sigmoid output to the derivative.
def sigmoid_derivative(output):
    return output * (1 - output)
```

```python
for i in range(100):

    # -------------------
    # Forward propagation
    # -------------------
    z1 = x * w1 + b1
    h = sigmoid(z1)

    z2 = h * w2 + b2
    y_pred = sigmoid(z2)

    # -------------------
    # Loss
    # -------------------
    loss = 0.5 * (y - y_pred) ** 2

    # -------------------
    # Backpropagation
    # -------------------

    # Output layer
    dL_dypred = y_pred - y
    dypred_dz2 = sigmoid_derivative(y_pred)

    dw2 = dL_dypred * dypred_dz2 * h
    db2 = dL_dypred * dypred_dz2

    # Hidden layer
    dz2_dh = w2
    dh_dz1 = sigmoid_derivative(h)

    dw1 = (
        dL_dypred
        * dypred_dz2
        * dz2_dh
        * dh_dz1
        * x
    )

    db1 = (
        dL_dypred
        * dypred_dz2
        * dz2_dh
        * dh_dz1
    )

    # -------------------
    # Update parameters
    # -------------------
    w1 = w1 - lr * dw1
    b1 = b1 - lr * db1

    w2 = w2 - lr * dw2
    b2 = b2 - lr * db2

    # -------------------
    # Print progress
    # -------------------
    if (i + 1) % 10 == 0:
        print(
            f"Iteration {i+1}: "
            f"Prediction={y_pred:.4f}, "
            f"Loss={loss:.6f}"
        )
```
