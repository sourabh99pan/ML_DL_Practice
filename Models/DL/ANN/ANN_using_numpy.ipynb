```python
import numpy as np

# Input and actual output
x = 2.0
y = 1.0

# Hidden layer
w1 = 0.5
b1 = 0.0

# Output layer
w2 = 1.0
b2 = 0.0

# Learning rate
lr = 0.1
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

```python
#Notice we're passing the sigmoid output to the derivative.
def sigmoid_derivative(output):
    return output * (1 - output)
```

```python
#h is calculated by passing the weighted sum of inputs to the sigmoid function.
z1 = x * w1 + b1
h = sigmoid(z1)
```

```python
#y_pred is calculated by passing the weighted sum of hidden layer output to the sigmoid function.
z2 = h * w2 + b2
y_pred = sigmoid(z2)
```

```python
print("z1:", z1)
print("Hidden output:", h)
print("z2:", z2)
print("Prediction:", y_pred)
```

```python
#loss is calculated using the Mean Squared Error (MSE) formula.
loss = 0.5 * (y - y_pred) ** 2

print("Loss:", loss)
```

```python
dL_dypred = y_pred - y

dypred_dz2 = sigmoid_derivative(y_pred)

dz2_dw2 = h

dw2 = dL_dypred * dypred_dz2 * dz2_dw2
print("Gradient w2:", dw2)
```

```python
dz2_dh = w2
dh_dz1 = sigmoid_derivative(h)
dz1_dw1 = x

dw1 = (
    dL_dypred
    * dypred_dz2
    * dz2_dh
    * dh_dz1
    * dz1_dw1
)
print("Gradient w1:", dw1)
```

```python
w2 = w2 - lr * dw2
w1 = w1 - lr * dw1
print("Updated w2:", w2)
print("Updated w1:", w1)
```

```python
z1 = x * w1 + b1
h = sigmoid(z1)

z2 = h * w2 + b2
y_pred = sigmoid(z2)

loss = 0.5 * (y - y_pred) ** 2

print("New prediction:", y_pred)
print("New loss:", loss)
```
