```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": [" Amit ", "Rahul", "Priya", "Rahul", "Neha "],
    "Age": [25, np.nan, 28, 32, 35],
    "Salary": [40000, 65000, 55000, 65000, np.nan],
    "City": ["Indore", "Mumbai", "Delhi", "Mumbai", " Indore "]
})
```

```python
df.describe()
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())
df.head()
```

```python
df.duplicated().sum()
```

```python
df["Name"] = df["Name"].str.strip()
df.head()
```

```python
df["Age"] = df["Age"].astype("int64")
df.dtypes
```
