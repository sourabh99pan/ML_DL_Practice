```python

import pandas as pd
df = pd.DataFrame({
    "Name": [" Amit ", "Rahul", "Priya", "Rahul", "Neha "],
    "Age": [25, 27, 28, 32, 35],
    "Salary": [40000, 65000, 55000, 65000, 75000],
    "City": ["Indore", "Mumbai", "Delhi", "Mumbai", " Indore "],
    "Education":["Graduate", "Post Graduate", "Graduate", "PhD", "Post Graduate"]
})

print(df)
```

```python
df = pd.get_dummies(df, columns=["City"])
print(df)
```

```python
from sklearn.preprocessing import OrdinalEncoder

encoder = OrdinalEncoder()

df["Education"] = encoder.fit_transform(
    df[["Education"]]
)
print(df)
```
