```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "EmployeeID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Name": [
        " Rahul ",
        "Priya",
        "Amit",
        "Neha",
        "Karan",
        "Rahul",
        "Sneha",
        "Vikas",
        "Priya",
        "Arjun"
    ],
    "Age": [28, 35, np.nan, 42, 29, 28, 31, np.nan, 35, 52],
    "City": [
        "Indore", "Mumbai", "Delhi", "Indore", "Pune",
        "Indore", "Mumbai", "Delhi", "Mumbai", "Pune"
    ],
    "Education": [
        "Bachelor", "Master", "PhD", "Bachelor", "Master",
        "Bachelor", "Master", "PhD", "Master", "Bachelor"
    ],
    "Experience": [3, 10, 15, 18, 4, 3, 7, 12, 10, 25],
    "MonthlySalary": [
        40000, 65000, 90000, 75000, 45000,
        40000, 60000, 85000, 65000, 500000
    ],
    "Bonus": [
        5000, 8000, 10000, np.nan, 4000,
        5000, 7000, 12000, 8000, 50000
    ],
    "JoiningDate": [
        "2020-01-10",
        "2018-06-15",
        "2015-03-20",
        "2017-08-01",
        "2021-02-10",
        "2020-01-10",
        "2019-05-12",
        "2016-11-25",
        "2018-06-15",
        "2010-01-05"
    ],
    "Department": [
        "QA", "Dev", "Data", "QA", "Dev",
        "QA", "Data", "Dev", "Dev", "Data"
    ]
})
```

```python
dup = df.duplicated().sum()
print(dup)
```

```python
df["Name"]=df["Name"].str.strip()
df["Education"] = df["Education"].str.strip()
df["City"] = df["City"].str.strip()
```

```python
import matplotlib.pyplot as plt

# Quick histogram of a column
df['Age'].hist(bins=10, edgecolor='black')
plt.title('Distribution of Age')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
df["Age"]  = df["Age"].fillna(df["Age"].median())
```

```python
import matplotlib.pyplot as plt

# Quick histogram of a column
df['Bonus'].hist(bins=10, edgecolor='black')
plt.title('Distribution of Bonus')
plt.xlabel('Bonus')
plt.ylabel('Frequency')
plt.show()
```

```python
df["Bonus"]  = df["Bonus"].fillna(df["Bonus"].median())
print(df)
```

```python
df["TotalIncome"] = df["MonthlySalary"] + df["Bonus"]
print(df)
```

```python
df["SalaryPerExperience"] = df["TotalIncome"] / df["Experience"]
print(df)
```

```python
df =  pd.get_dummies(df, columns=["City"])
```

```python
from sklearn.preprocessing import OrdinalEncoder

encoder = OrdinalEncoder(
    categories=[["Bachelor", "Master", "PhD"]]
)

df["Education"] = encoder.fit_transform(
    df[["Education"]]
)
print(df)
```

```python
Q1 = df["TotalIncome"].quantile(0.25)
Q3 = df["TotalIncome"].quantile(0.75)

IQR= Q3-Q1
print(IQR)
```

```python
lower1 = Q1-1.5*IQR
upper1 = Q3+1.5*IQR
outliers = df[(df["TotalIncome"]<lower1) | (df["TotalIncome"]>upper1)]
print(outliers)
```

```python
df['TotalIncome'] = df['TotalIncome'].clip(upper=upper1)
```

```python
df["AgeGroup"] = pd.cut(df["Age"], bins=[0,29,39,49,100],
			labels = ["young","Adult","Middle","Senior"])
print(df)
```

```python
from sklearn.preprocessing import OrdinalEncoder

encoder = OrdinalEncoder()

df["Department"] = encoder.fit_transform(
    df[["Department"]]
)
print(df)
```

```python
df["JoiningDate"] = pd.to_datetime(df["JoiningDate"])
```

```python
df["JoiningYear"] = df["JoiningDate"].dt.year
```

```python
df["JoiningMonth"] = df["JoiningDate"].dt.month
print(df)
```

```python
linear_corr = df.corr(numeric_only=True)['Department'].abs().sort_values(ascending=False)
print(linear_corr.head(10))
```

```python
clean_corr = linear_corr.drop(labels=['Department', 'EmployeeID'], errors='ignore')

print("Top Predictors for Department:")
print(clean_corr.head(10))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
clean_corr.head(8).plot(kind='barh', color='teal').invert_yaxis()
plt.title('Correlation with Department')
plt.xlabel('Absolute Correlation Coefficient')
plt.show()
```
