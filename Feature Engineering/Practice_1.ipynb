```python
import pandas as pd

df = pd.DataFrame({
    "Age": [22, 27, 34, 42, 55],
    "MonthlySalary": [30000, 40000, 60000, 80000, 100000],
    "Bonus": [5000, 6000, 10000, 15000, 20000],
    "Experience": [1, 3, 8, 15, 25]
})
```

```python
df["AnnualSalary"] = df["MonthlySalary"] * 12 
```

```python
df["TotalMonthlyIncome"] = df["MonthlySalary"] + df["Bonus"]
```

```python
df["AnnualIncome"] = df["TotalMonthlyIncome"] * 12
```

```python
df["SalaryPerExperience"] = df["AnnualSalary"] / df["Experience"]
```

```python
df["AgeExperience"] = df["Age"]*df["Experience"]
```
