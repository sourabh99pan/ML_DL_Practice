```python
import pandas as pd
df = pd.DataFrame({'Number':[0,10, 12, 14, 15, 16, 18, 20, 100,150,200]})
```

```python
Q1 = df["Number"].quantile(0.25)
Q3 = df["Number"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
```

```python
outliers = df[
    (df["Number"] < lower) |
    (df["Number"] > upper)
]

print("Outliers:")
print(outliers)
```
