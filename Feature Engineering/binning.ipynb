```python
import pandas as pd
df = pd.DataFrame({
    "Age": [18, 22, 29, 30, 35, 40, 49, 50, 65],
    "Salary": [25000, 30000, 40000, 50000, 60000, 70000, 80000, 100000, 150000]
})
```

```python
df["AgeGroup"] = pd.cut(df["Age"], bins=[0,29,39,49,100],
			labels = ["young","Adult","Middle","Senior"])
```

```python
df["SalaryGroup"] = pd.cut(df["Salary"], bins=[0,39999,69999,99999,150000],
		labels=["low", "medium", "high", "very high"])
```
